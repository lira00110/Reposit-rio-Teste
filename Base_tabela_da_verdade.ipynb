```python
import itertools as it #Importa a biblioteca itertools (ler mais sobre a biblioteca)

def truth_table_and(expression):
# O Método truth_table_and() é um método para a expressão (expression que será utilizada dentro de eval() para a conversão de string para código literal)

  true_false = [[0,1], [0,1]]
# Os valores são utilizados apenas para P e Q

  saida = []
# A saída é uma matriz que será utilizada para agrupar os valores de P, Q, not P e resultado

  for valores in it.product(*true_false):
  # Cria um for para as valores de 0,1 (que são os valores da variável true_false)

    P,Q = valores
    #P e Q recebem os valores da variável true_false

    not_P = not P
    #Cria um not P que será utilizado posteriormente para a tabela da verdade

    resultado = eval(expression)
    # A variável resultado, converte o valor da expressão utilizada dentro do método truth_table_and() para código literal com o eval() (ler mais sobre o método eval())

    saida.append((valores, not_P, resultado))
    # Agrupa P e Q, not_P e resultado na variavel saida que é uma tupla destes elementos


  print("{:<5} {:<5} {:<8} {:<10}".format('P', 'Q', 'not P', expression))
  # Neste trecho as strings P, Q not P e expression serão formatadas para o número de caracteres de distância referenciados em {:<(distância)}

  for valores, not_P, resultado in saida:
    #A partir dos valores dentro da tupla...

    print("{:<5} {:<5} {:<8} {:<10}".format(*valores, int(not_P), int(resultado)))
    # Ele pega os valores e interage para cada um dos itens. Como not_P é um valor booleano e ele está dentro da variável resultado (que foi convertida para código), ele converte o valor de True e False para 0,1 respectivamente para aparecerem dentro da tabela como valores.

truth_table_and("not P and Q")
# A expressão é chamada e convertida de strings para código
```
